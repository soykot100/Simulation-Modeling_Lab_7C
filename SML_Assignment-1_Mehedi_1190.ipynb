#Task 1: Scalar & Araay **Operation**

```python
import numpy as np
scalar = 4
arr = np.array([1, 3.5, 2, -6, 5])
print("Scalar value:",scalar)
print("Array:",arr)
print()

add_value = scalar + arr
sub_value = scalar - arr
mul_value = scalar * arr
div_value = scalar / arr
pow_value = scalar ** arr

print("Addition :", add_value)
print("Subtraction :", sub_value)
print("Multiplication:", mul_value)
print("Division :", div_value)
print("power:", pow_value)
```

### #Trigonometric function

```python
sin_val = np.sin(arr)
cos_val = np.cos(arr)
tan_val = np.tan(arr)

print("sin:", sin_val)
print("cos:", cos_val)
print("tan:", tan_val)

#Inverse FUnction
arr_clip = np.clip(arr/10, -1, 1)

insin_val = np.arcsin(arr_clip)
incos_val = np.arccos(arr_clip)
inatn_val = np.arctan(arr)

print("Inverse sin:", insin_val)
print("Inverse cos:", incos_val)
```

## Expotential function

```python
exp_val = np.exp(arr)
print("Exponential Value",exp_val)
```

# Logarithm Function

```python
log_val = np.log(arr)
lon2_val = np.log2(arr)
print("Log Value",log_val)
print("Log2 Value",lon2_val)
```

## Other MathMatical Function

```python
abs_val = np.abs(arr)
sqrt_val = np.sqrt(arr)
ceil_val = np.ceil(arr)
floor_val = np.floor(arr)
std_val = np.std(arr)
var_val = np.var(arr)
min_val = np.min(arr)
max_val = np.max(arr)
sum_val = np.sum(arr)
prod_val = np.prod(arr)
mean_val = np.mean(arr)
median_val = np.median(arr)
sort_val = np.sort(arr)

print("Absolute Value",abs_val)
print("Square Root Value",sqrt_val)
print("Ceil Value",ceil_val)
print("Floor Value",floor_val)
print("Standard Deviation Value",std_val)
print("Variance Value",var_val)
print("Minimum Value",min_val)
print("Maximum Value",max_val)
print("Sum Value",sum_val)
print("Product Value",prod_val)
print("Mean Value",mean_val)
print("Median Value",median_val)
print("Sorted Value",sort_val)
```

#Task 2: 3*2 Matrix Calculation

```python
matrix = np.array([[2, 3],[4, 5],[6, 8.5]])
print("Matrix:")
print(matrix)

# Trigonometric functions on matrix
sin_mat = np.sin(matrix)
cos_mat = np.cos(matrix)
tan_mat = np.tan(matrix)

print("sin(matrix):")
print(sin_mat)
print()
print("cos(matrix):")
print(cos_mat)
print()
print("tan(matrix):")
print(tan_mat)

# Clip values for arcsin and arccos (valid range: -1 to 1)
matrix_clipped = np.clip(matrix/10, -1, 1)

arcsin_mat = np.arcsin(matrix_clipped)
arccos_mat = np.arccos(matrix_clipped)
arctan_mat = np.arctan(matrix)

print("arcsin(matrix) - after clipping:")
print(arcsin_mat)
print()
print("arccos(matrix) - after clipping:")
print(arccos_mat)
print()
print("arctan(matrix):")
print(arctan_mat)

# Exponential and logarithmic functions
exp_mat = np.exp(matrix)

# Adding small value to avoid log(0) - though no zero values here
log_mat = np.log(matrix + 0.01)

log2_mat = np.log2(matrix + 0.01)

print("Exponential (e^matrix):")
print(exp_mat)
print()
print("Natural log (log base e):")
print(log_mat)
print()

print("Log base 2:")
print(log2_mat)

# Rounding functions
round_mat = np.round(matrix)
floor_mat = np.floor(matrix)
ceil_mat = np.ceil(matrix)

print("Round (nearest integer):")
print(round_mat)
print()
print("Floor (round down):")
print(floor_mat)
print()
print("Ceil (round up):")
print(ceil_mat)
```

## #TASK 3: TASK 3: VECTOR FUNCTIONS ON 2×3 MATRIX

```python
vec_matrix = np.array([[5, 1],[2, -8],[4, 6.3]])

print("Original 3×2 Matrix:")
print(matrix)
print()

# Max & Min (with indices)
row_max = np.max(vec_matrix, axis=1)
row_min = np.min(vec_matrix, axis=1)
row_max_idx = np.argmax(vec_matrix, axis=1)
row_min_idx = np.argmin(vec_matrix, axis=1)

print("Row-wise max:", row_max)
print("Indices of row-wise max:", row_max_idx)
print()
print("Row-wise min:", row_min)
print("Indices of row-wise min:", row_min_idx)

# Column-wise max and min
col_max = np.max(vec_matrix, axis=0)
col_min = np.min(vec_matrix, axis=0)
col_max_idx = np.argmax(vec_matrix, axis=0)
col_min_idx = np.argmin(vec_matrix, axis=0)

print("Column-wise max:", col_max)
print("Indices of column-wise max:", col_max_idx)
print()
print("Column-wise min:", col_min)
print("Indices of column-wise min:", col_min_idx)

# Vector length (Euclidean norm)
row_length = np.linalg.norm(vec_matrix, axis=1)
col_length = np.linalg.norm(vec_matrix, axis=0)
print("Vector length (row-wise):", row_length)
print("Vector length (column-wise):", col_length)

# Overall max and min
overall_max = np.max(vec_matrix)
overall_min = np.min(vec_matrix)
print("Overall max:", overall_max)
print("Overall min:", overall_min)

# Sorted version
row_sorted_asc = np.sort(vec_matrix, axis=1)
row_sorted_desc = np.sort(vec_matrix, axis=1)[:, ::-1]
col_srt_asc = np.sort(vec_matrix, axis=0)
col_srt_desc = np.sort(vec_matrix, axis=0)[:, ::-1]
print("Row-wise sorted (ascending):")
print(row_sorted_asc)
print()
print("Row-wise sorted (descending):")
print(row_sorted_desc)
print()
print("Column-wise sorted (ascending):")
print(col_srt_asc)
print()
print("Column-wise sorted (descending):")
print(col_srt_desc)
```

#Mean, Midain, Std function

```python
# Median, Mean, Standard Deviation
mean_all = np.mean(vec_matrix)
median_all = np.median(vec_matrix)
std_all = np.std(vec_matrix)
print("Mean of all elements:", mean_all)
print("Median of all elements:", median_all)
print("Standard Deviation of all elements:", std_all)
```
