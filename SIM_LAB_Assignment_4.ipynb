#Task-2 : Sample Means, Variances, and Mean Difference

```python
data1 = [18.4, 19.1, 17.9, 18.7, 18.3]
data2 = [20.2, 20.5, 20.1, 20.3, 20.4]

# Convert to numpy arrays
group1 = np.array(data1)
group2 = np.array(data2)

# Sample means
mean1 = np.mean(group1)
mean2 = np.mean(group2)

# Sample variances
var1 = np.var(group1, ddof=1)
var2 = np.var(group2, ddof=1)
print("""Degrees of freedom (dof) represent the number
of independent values that can vary in a
statistical calculation.""")

print(f"\nGroup 1 Data: {data1}")
print(f"Group 2 Data: {data2}")
print(f"\nSample Mean 1: {mean1:.2f}")
print(f"Sample Mean 2: {mean2:.2f}")
print(f"Sample Variance 1: {var1:.2f}")
print(f"Sample Variance 2: {var2:.2f}")

# Sample sizes
n1 = len(group1)
n2 = len(group2)

# Mean difference
mean_diff = mean1 - mean2
print(f"\nMean Difference: {mean_diff:.2f}")
print(f"Sample Size 1: {n1}")
print(f"Sample Size 2: {n2}")
```

# TASK 3: T-Statistic, Degrees of Freedom, P-Value

```python

t_stat, p_value = stats.ttest_ind(group1, group2, equal_var=False)


numerator = (var1/n1 + var2/n2)**2
denominator = ((var1/n1)**2)/(n1-1) + ((var2/n2)**2)/(n2-1)
df = numerator / denominator

print(f"\nT-Statistic: {t_stat:.2f}")
print(f"Degrees of Freedom: {df:.0f}")
print(f"P-Value: {p_value:.4f}")


t_stat_pooled, p_value_pooled = stats.ttest_ind(group1, group2, equal_var=True)
df_pooled = n1 + n2 - 2

print(f"\nT-Statistic (Pooled Variance): {t_stat_pooled:.2f}")
print(f"Degrees of Freedom (Pooled Variance): {df_pooled:.0f}")
print(f"P-Value (Pooled Variance): {p_value_pooled:}")

```

# TASK 4: 95% Confidence Interval

```python
# Standard error of the difference
se_diff = np.sqrt(var1/n1 + var2/n2)

# Critical t-value for 95% CI (two-tailed)
alpha = 0.05
t_critical = stats.t.ppf(1 - alpha/2, df)

# Margin of error
margin_error = t_critical * se_diff

# Confidence Interval
ci_lower = mean_diff - margin_error
ci_upper = mean_diff + margin_error


print(f"\nMean Difference: {mean_diff:.4f}")
print(f"Standard Error of Difference: {se_diff:.4f}")
print(f"Critical t-value (df={df:.2f}, α/2=0.025): {t_critical:.4f}")
print(f"Margin of Error: {margin_error:.4f}")

print(f"\n95% Confidence Interval: ({ci_lower:.4f}, {ci_upper:.4f})")

print(f"\nInterpretation:")
print(f"We are 95% confident that the true difference between")
print(f"population means lies between {ci_lower:.4f} and {ci_upper:.4f}.")
```
