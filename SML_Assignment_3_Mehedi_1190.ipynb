```python
import numpy as np
import matplotlib.pyplot as plt

print("Libraries loaded")
```

```python
numbers = np.random.rand(10)

print("10 Random Numbers:")
print(numbers)

```

# Task-1 -Draw histogram to see distribution

```python
plt.figure(figsize=(8, 4))
plt.hist(numbers, bins=5, color='skyblue', edgecolor='black')
plt.title('Histogram: 10 Random Numbers (0-1)')
plt.xlabel('Value')
plt.ylabel('Frequency')
plt.grid(True)
plt.show()
```

#Task 2: Normal Distribution (500 numbers)

```python
normal_numbers = np.random.normal(0, 1, 500)

print("First 10 numbers:", normal_numbers[:10])
print("Mean:", normal_numbers.mean())
print("Standard Deviation:", normal_numbers.std())
```

```python
# Draw histogram
plt.figure(figsize=(8, 4))
plt.hist(normal_numbers, bins=30, color='lightgreen', edgecolor='black')
plt.title('Histogram: 500 Normal Numbers (Mean=0)')
plt.xlabel('Value')
plt.ylabel('Frequency')
plt.grid(True)
plt.show()
```

#Task 3: Random Integers (3×3 Matrix)

```python
np.random.seed(10)

# Create 3x3 matrix with numbers
matrix_with_seed = np.random.randint(1, 51, size=(3, 3))

print("Matrix WITH Seed (always same):")
print(matrix_with_seed)

matrix_without_seed1 = np.random.randint(1, 51, size=(3, 3))
matrix_without_seed2 = np.random.randint(1, 51, size=(3, 3))

print("Without Seed (Run 1):")
print(matrix_without_seed1)

print("\nWithout Seed (Run 2 - Different!):")
print(matrix_without_seed2)
```

#Task 4- One Sample T-Test

```python
data = [5.2, 5.4, 5.1, 5.6, 5.3, 5.8]
hypothesized_mean = 5.5

print("Our Data:", data)
print("Hypothesized Mean:", hypothesized_mean)
print("Total numbers:", len(data))
```

```python
from scipy import stats

# Do t-test
t_stat, p_value = stats.ttest_1samp(data, hypothesized_mean)

print("\n--- T-Test Results ---")
print("T-Statistic:", round(t_stat, 4))
print("P-Value:", round(p_value, 4))
```

```python
# Simple graph
plt.figure(figsize=(10, 4))

# Show each data point
plt.scatter(range(len(data)), data, s=100, color='blue', label='Data')

# Show data mean (red line)
plt.axhline(y=np.mean(data), color='red', linewidth=2, label=f'Mean = {np.mean(data):.2f}')

# Show hypothesized mean (green dashed line)
plt.axhline(y=hypothesized_mean, color='green', linestyle='--', linewidth=2, label=f'Hypothesized = {hypothesized_mean}')

plt.title('Data Points vs Mean Comparison')
plt.xlabel('Index')
plt.ylabel('Value')
plt.legend()
plt.grid(True)
plt.show()
```
